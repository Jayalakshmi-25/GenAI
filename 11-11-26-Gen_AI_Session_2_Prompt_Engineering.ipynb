```python
import fitz

def extract_pdf_text(pdf_path):
    text = ""

    with fitz.open(pdf_path) as pdf:
        for page in pdf:
            text += page.get_text("text") + "\n"

    return text


# Resume PDF file
pdf_path = "/content/Pramod _Devops_Resume.pdf"
```

```python
resume_text=extract_pdf_text(pdf_path)
```
