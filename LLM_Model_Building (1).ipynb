```python
# ============================================================
# STEP 1: INSTALL PACKAGES (Google Colab)
# ============================================================

# !pip install -U ragas langchain-google-genai langchain-core pandas

# ============================================================
# STEP 2: IMPORT LIBRARIES
# ============================================================

import os
import asyncio
import pandas as pd
from getpass import getpass

from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings
)

from ragas import evaluate, EvaluationDataset, SingleTurnSample

from ragas.metrics import (
    Faithfulness,
    ResponseRelevancy,
    LLMContextPrecisionWithReference,
    LLMContextRecall
)

# RAGAS adapters for LangChain models
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper

# ============================================================
# STEP 4: INITIALIZE GEMINI LLM
# ============================================================

llm = ChatGoogleGenerativeAI(
    model="gemini-4-hydrogen",
    google_api_key=google_api_key,
    temperature=0
)

embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    google_api_key=google_api_key
)

# Convert LangChain models to RAGAS-compatible models
ragas_llm = LangchainLLMWrapper(llm)

ragas_embeddings = LangchainEmbeddingsWrapper(
    embedding_model
)


# ============================================================
# STEP 5: CREATE LARGE BANKING KNOWLEDGE BASE
# ============================================================

context1 = """
ABC Bank is a multinational financial institution founded in
1995. The bank operates across India, Singapore, the United
Kingdom, and the United States.

The bank provides retail banking, corporate banking, investment
banking, credit cards, personal loans, home loans, and digital
payment services.

ABC Bank serves approximately 25 million customers worldwide
and operates more than 1,200 branches.

The bank launched its digital banking transformation program
in 2020 to modernize legacy applications and improve customer
experience.

The transformation program introduced cloud infrastructure,
microservices architecture, real-time analytics, and artificial
intelligence capabilities.

Microsoft Azure was selected as the primary cloud platform
for enterprise data engineering and analytics workloads.
"""

context2 = """
ABC Bank implemented a modern data engineering platform
using Azure Data Factory, Azure Databricks, Azure Data Lake
Storage Gen2, Delta Lake, and Power BI.

Azure Data Factory is responsible for extracting data from
multiple enterprise systems including SQL Server, Oracle,
REST APIs, CSV files, and third-party banking applications.

Azure Databricks performs data transformation using PySpark
and Delta Lake.

The platform follows Medallion Architecture with three layers:

Bronze Layer:
Stores raw data extracted from source systems without
major transformations.

Silver Layer:
Performs data cleansing, duplicate removal, null handling,
schema validation, and standardization.

Gold Layer:
Creates business-ready datasets for customer analytics,
fraud detection, regulatory reporting, and dashboards.

Power BI connects to curated Gold datasets to generate
interactive dashboards for business stakeholders.
"""

context3 = """
ABC Bank processes approximately 50 million financial
transactions every day.

Transaction data includes transaction ID, customer ID,
account number, transaction amount, transaction timestamp,
merchant category, transaction location, and payment method.

The bank uses Apache Spark on Azure Databricks to process
large transaction datasets.

Incremental data ingestion is implemented using watermark
columns and change data capture techniques.

Delta Lake MERGE operations are used to implement upserts
and maintain updated transaction records.

The bank uses partitioning, data skipping, and optimized
Delta tables to improve analytical query performance.

Daily processing pipelines are orchestrated through Azure
Data Factory and Databricks workflows.
"""

context4 = """
ABC Bank developed a fraud detection system using machine
learning and real-time transaction monitoring.

The fraud detection platform analyzes transaction frequency,
transaction amount, geographic location, device information,
and historical customer spending behavior.

Machine learning models classify transactions as legitimate
or potentially fraudulent.

The system uses historical labeled transaction data to
train fraud detection models.

The fraud detection team evaluates models using precision,
recall, F1-score, and ROC-AUC.

Suspicious transactions are flagged for further investigation.

The fraud detection system helps reduce financial losses
and improves transaction security.
"""

context5 = """
ABC Bank implemented a Retrieval-Augmented Generation
application to assist customer support representatives.

The RAG system retrieves information from banking policy
documents, product manuals, loan guidelines, and customer
support knowledge articles.

Documents are divided into smaller chunks before generating
vector embeddings.

The embeddings are stored in a vector database.

When a customer asks a question, the system retrieves
relevant document chunks using semantic similarity search.

The retrieved documents are provided to a large language
model to generate a grounded answer.

The bank evaluates the RAG application using faithfulness,
answer relevancy, context precision, and context recall.

Faithfulness measures whether generated statements are
supported by retrieved documents.

Answer relevancy measures whether the generated response
addresses the original customer question.

Context precision evaluates the ranking of relevant
retrieved information.

Context recall measures whether the retrieved information
covers the reference answer.
"""


# ============================================================
# STEP 6: COMBINE RETRIEVED DOCUMENTS
# ============================================================

retrieved_contexts = [
    context1,
    context2,
    context3,
    context4,
    context5
]

print("Total Retrieved Documents:", len(retrieved_contexts))

for i, context in enumerate(retrieved_contexts, start=1):
    print(f"Document {i} Characters:", len(context))


# ============================================================
# STEP 7: DEFINE USER QUESTION
# ============================================================

question = """
Explain how ABC Bank uses Azure Data Factory,
Azure Databricks, Delta Lake, Medallion Architecture,
and Power BI in its data engineering platform.
"""


# ============================================================
# STEP 8: GENERATE ANSWER USING GEMINI
# ============================================================

combined_context = "\n\n".join(retrieved_contexts)

prompt = f"""
You are a banking data engineering expert.

Answer the user's question using only the supplied context.

Do not introduce information that is not present in
the retrieved documents.

CONTEXT:
{combined_context}

QUESTION:
{question}

ANSWER:
"""

response = llm.invoke(prompt)

generated_answer = response.content

# Handle content-block responses if returned by the model
if isinstance(generated_answer, list):
    generated_answer = " ".join(
        block.get("text", "")
        for block in generated_answer
        if isinstance(block, dict)
    )

print("\n========== GENERATED ANSWER ==========\n")
print(generated_answer)


# ============================================================
# STEP 9: DEFINE GROUND TRUTH / REFERENCE ANSWER
# ============================================================

reference_answer = """
ABC Bank uses Azure Data Factory to ingest banking data
from SQL Server, Oracle, REST APIs, CSV files, and
third-party banking applications.

Azure Databricks uses PySpark and Delta Lake to process
and transform large banking datasets.

The platform follows Medallion Architecture.

The Bronze layer stores raw source data.

The Silver layer performs cleansing, duplicate removal,
null handling, and schema validation.

The Gold layer produces curated business datasets for
customer analytics, fraud detection, regulatory reporting,
and business intelligence.

Delta Lake supports MERGE operations for incremental
updates and maintains optimized analytical tables.

Power BI connects to Gold datasets to create interactive
dashboards for business stakeholders.
"""


# ============================================================
# STEP 10: CREATE RAGAS EVALUATION SAMPLE
# ============================================================

sample = SingleTurnSample(
    user_input=question,
    retrieved_contexts=retrieved_contexts,
    response=generated_answer,
    reference=reference_answer
)

dataset = EvaluationDataset(
    samples=[sample]
)


# ============================================================
# STEP 11: INITIALIZE ALL FOUR RAGAS METRICS
# ============================================================

metrics = [
    Faithfulness(llm=ragas_llm),

    ResponseRelevancy(
        llm=ragas_llm,
        embeddings=ragas_embeddings
    ),

    LLMContextPrecisionWithReference(
        llm=ragas_llm
    ),

    LLMContextRecall(
        llm=ragas_llm
    )
]


# ============================================================
# STEP 12: PRINT METRIC NAMES
# ============================================================

print("\n========== METRICS ==========\n")

for metric in metrics:
    print(type(metric).__name__)


# ============================================================
# STEP 13: RUN RAGAS EVALUATION
# ============================================================

result = evaluate(
    dataset=dataset,
    metrics=metrics,
    raise_exceptions=True
)


# ============================================================
# STEP 14: PRINT ALL METRIC SCORES
# ============================================================

print("\n========== RAGAS EVALUATION RESULTS ==========\n")

for metric_name, score in result.items():
    print(f"{metric_name}: {score:.4f}")


# ============================================================
# STEP 15: DISPLAY RESULTS AS DATAFRAME
# ============================================================

df = result.to_pandas()

print("\n========== COMPLETE EVALUATION ==========\n")

display(df)


# ============================================================
# STEP 16: CREATE METRIC SUMMARY
# ============================================================

metric_scores = dict(result.items())

summary = pd.DataFrame({
    "Metric": list(metric_scores.keys()),
    "Score": list(metric_scores.values())
})

print("\n========== METRIC SUMMARY ==========\n")

display(summary)
```

```python
Metric	What it measures
Faithfulness	Is the answer supported by the retrieved context?
Answer Relevancy	Does the answer address the user's question?
Context Precision	Are relevant documents ranked near the top?
Context Recall	Did retrieval capture the information needed to answer?
Answer Correctness	Is the generated answer factually correct compared with the reference?
Answer Similarity	How semantically similar is the answer to the reference?
```

```python
import math

probabilities = [0.8, 0.9, 0.7, 0.6]

negative_log_likelihood = -sum(
    math.log(p) for p in probabilities
) / len(probabilities)

perplexity = math.exp(negative_log_likelihood)

print("Negative Log Likelihood:", negative_log_likelihood)
print("Perplexity:", perplexity)
```

```python
Perplexity : the lower the value the better the model should be
```
