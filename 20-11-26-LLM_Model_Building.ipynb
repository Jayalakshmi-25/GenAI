```python
### Load pdf,parse pdf and extract text
from langchain_community.document_loaders import PyPDFLoader
pdf_path="/content/Generative-AI-and-LLMs-for-Dummies.pdf"
loader=PyPDFLoader(pdf_path)
documents=loader.load()
len(documents)
final_text=" "
for pagenumber,pagecontent in enumerate(documents):
  #print(pagenumber,pagecontent.page_content)
  final_text=final_text+pagecontent.page_content
len(final_text),len(final_text.split())
```

```python
len(final_text)
```

```python
### Chunking
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200,length_function=len,separators=["\n\n", "\n", " ", ""])
chunks=text_splitter.split_text(final_text)
```

```python
from langchain_chroma import Chroma

vector_db = Chroma.from_texts(
    texts=chunks[0:30],
    embedding=embedding_model,
    collection_name="generative_ai_book",
    persist_directory="./chroma_db"
)

print("Total vectors stored:", vector_db._collection.count())
```

```python
import chromadb

# Create persistent ChromaDB client
client = chromadb.PersistentClient(path="./chroma_db")

# Create collection
collection = client.get_or_create_collection(
    name="generative_ai_book",
    metadata={"hnsw:space": "cosine"}
)

# Store existing chunks and Gemini embeddings
collection.upsert(
    ids=[f"chunk_{i}" for i in range(len(chunks[0:30]))],
    documents=chunks[0:30],
    embeddings=embeddings
)

print("Total vectors stored:", collection.count())
```
